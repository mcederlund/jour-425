# requests
The simplest way to pull data from the internet.

You can request text files, csv, json and even binary formats like xlsx, jp and pdf, to name a few.


```python
# fetch the data
import requests
# sample url we want to pull (a collection of robert frost poems)
poem_url = "https://www.gutenberg.org/cache/epub/59824/pg59824.txt"
# get the response
response = requests.get(poem_url)
# did it work? http code 200 means yes!
response.status_code
# get the content
poem = response.text
# print the response
print(poem)
# save it
with open('poem.txt','w') as outfile:
  outfile.write(poem)
```

```python
# write json to file
import json
# load json
speed_camera_violation = json.loads(ff_response.text)
# open json file
with open('speed_camera_violation.json','w') as outfile:
  json.dump(speed_camera_violation,outfile)
```

```python
# import to csv
import csv
# change from json to csv
speed_camera_violation = 'https://data.cityofchicago.org/resource/hhkd-xvj4.csv?$limit=500000'
# get the csv version
ff_response = requests.get(speed_camera_violation, headers=headers)
speed_camera_violation_csv_txt = ff_response.text
# open csv file
with open('speed_camera_violation.csv','w') as outfile:
  outfile.write(speed_camera_violation_csv_txt)
```
